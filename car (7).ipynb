## 1. Introduction

The Vehicle Sales and Pricing Analysis project focuses on analyzing used-vehicle sales data using Python. The project aims to understand the factors that influence vehicle selling prices, identify sales trends, and compare actual selling prices with the Manheim Market Report (MMR) values.

The dataset contains information about vehicle manufacturers, models, manufacturing years, body types, transmission types, condition, odometer readings, market values, and actual selling prices.

Python libraries such as Pandas and NumPy were used for data manipulation, cleaning, and analysis, while Matplotlib and Seaborn were used to create visualizations. Through Exploratory Data Analysis (EDA), the project identifies patterns, correlations, and unusual values in the dataset.

### 2.4 Creating Derived Columns

Two new columns were created to compare actual vehicle selling prices with their market reference values.

Profit calculation:

1.PROFIT=SELLING PRICE−MMR

The PROFIT column represents the difference between the actual selling price and the market reference value.

2.PROFIT_STATUS, was created using the following conditions:

Profit: Selling price is greater than MMR.

Loss: Selling price is less than MMR.

Break-even: Selling price is equal to MMR.

This transformation helps identify how many vehicles sold above, below, or exactly at their market reference values.

These labels represent the price difference relative to MMR, not the seller's actual business profit, because purchase costs and operating expenses are not available in the dataset.

### 7.3 Anomalies and Data Quality Issues

Several unusual values and data quality issues were identified during the analysis.

Extreme odometer reading: The maximum odometer value was 999,999, which is unusually high compared with the average odometer reading of approximately 31,210. This value should be checked against the original source to determine whether it represents a genuine reading or a data-entry issue.


Inconsistent text encoding: Some categorical values in the INTERIOR column may contain incorrectly encoded characters, such as the unusual em-dash-like character observed during data inspection that was changed to green. These values should be standardized before grouping or visualization.

Limited geographic variation: The STATE column contains only one distinct value, "ca". Therefore, meaningful comparisons between different states cannot be performed using this dataset.

Identifier column: VIN contains unique vehicle identification numbers. Although useful for identifying individual records, VIN should generally not be treated as a numerical variable for correlation analysis.

These issues highlight the importance of validating data quality before drawing conclusions or developing predictive models.

DATA PREPROCESSING

```python
# FINDING THE NUMBER OF MISSING VALUES
df.isnull().sum()
```

```python
#For categorical columns, mode (most frequently occurring value) to fill the missing values.
categorical_columns = [
    'MAKE',
    'MODEL',
    'TRIM',
    'BODY',
    'TRANSMISSION',
    'COLOR',
    'INTERIOR'
]

for column in categorical_columns:
    df[column] = df[column].fillna(df[column].mode()[0])
```

```python
#For numerical column, used the median because it is less affected by extreme values than the mean.
df['CONDITION'] = df['CONDITION'].fillna(df['CONDITION'].median())
```

```python
# to find the duplicates and to remove
df.duplicated().sum()
df = df.drop_duplicates()
df.duplicated().sum()
```

```python
# replacing â€” with green colour in interior column
df['INTERIOR'] = df['INTERIOR'].replace('â€”', 'green')
df[df['INTERIOR'] == 'green']
df["COLOR"] = df["COLOR"].replace("—", "pink")
df
```

```python
# converting data type from object to datetime
df['SALED DATE'] = pd.to_datetime(df['SALED DATE'], utc=True)

df['SALED DATE'].dtype
```

```python
# new column price difference as profit created using existing columns
df['PROFIT'] = df['SELLING PRICE'] - df['MMR']
```

```python
# standardizing the make
df["MAKE"] = df["MAKE"].str.upper()
df
```

```python
# new column profit status is derived to know whether the status of profit column
profit_status = []

for profit in df['PROFIT']:
    if profit > 0:
        profit_status.append('Profit')
    elif profit < 0:
        profit_status.append('Loss')
    else:
        profit_status.append('Break-even')

df['PROFIT_STATUS'] = profit_status
```

```python
# filtering used to find the expensive cars
expensive_cars = df[df['SELLING PRICE'] > 80000]

expensive_cars
```

```python
# filtered using 2 condition selling price greater than 50,000 and condition greater or equal to 4
filtered_cars = df[
    (df['SELLING PRICE'] > 50000) &
    (df['CONDITION'] >= 4)
]

filtered_cars
```
